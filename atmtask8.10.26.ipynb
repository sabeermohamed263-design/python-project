```python
#leap year not
year=int(input("enter the year:"))
if year%4==0:                                    #leap year accuratta find panna 3 rule 4 aala divide aaganu 400 aala divide aahanu 
    if year%400==0 or year%100!= 0:              #and 100 aala divide aaha kudathu               
        print(year,"is leap year")              #(year%4==0,year%400==0 and year%100!=0)
    else:
        print(year,"is not leap year")
else:
    print(year,"is not leap year")
         
```
